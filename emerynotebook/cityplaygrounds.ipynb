```python
import pandas as pd

data = pd.read_csv("City_of_Pittsburgh_Playgrounds.csv")
%matplotlib inline

neighborhoods = data.groupby("neighborhood").agg(
    numParks = ("id", "count")
)

scores = neighborhoods.sort_values(
    by = "numParks", ascending = False
)

bestNeighborhood = scores.index[0]
print("Best Neighborhood:", bestNeighborhood)
```
