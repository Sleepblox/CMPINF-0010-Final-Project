```python
import pandas as pd

data = pd.read_csv("City_of_Pittsburgh_Playgrounds.csv")
%matplotlib inline

neighborhoods = data.groupby("neighborhood").agg(
    numParks = ("id", "count")
)

scores = neighborhoods.sort_values(
    by = "numParks", ascending = False
)

finalScores = scores.rank(pct= True) * 100

bestNeighborhood = finalScores.index[0]
print("Best Neighborhood:", bestNeighborhood)

finalScores.to_csv("../finalMetric/finalScoresPlaygrounds.csv")


import matplotlib.pyplot as plt

playgroundGraph = scores.head(15).plot.bar(
    title = "Top 15 Pittsburgh Neighborhoods by Number of Playgrounds",
    xlabel = "Neighborhood",
    ylabel = "Number of Playgrounds"
)
```
