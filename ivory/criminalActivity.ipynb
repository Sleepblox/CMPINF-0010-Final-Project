```python
import pandas as pd
crim = pd.read_csv("incidents_2024_sep2026(AllMergedTables).tsv", sep="\t", low_memory=False)
```

To find the best neighborhood in Pittsburgh, I decided to look at statistics for crime reports in various cities across Pittsburgh. I had some other ideas, including things like air quality reports, hospital ratings, or other health-related topics. However, I decided it would be a good choice to go with something safety-related, as our group decided that was an important statistic to look at. My metric looks at crime reports in the city, looking at which neighborhoods have the fewest crime reports, they will be rated higher than other neighborhoods. 

```python
dictionary = {}
for index, row in crim.iterrows():
    curNeighborhood = crim.at[index, 'Neighborhood']
    if curNeighborhood in dictionary:
        dictionary[curNeighborhood] = dictionary[curNeighborhood] + 1
    else:
        dictionary[curNeighborhood] = 1
print(dictionary)
```

Here, each report for each neighborhood is counted. However, this data is out of order, so let's sort it.

```python
series = pd.Series(dictionary)
sorted_series = series.sort_values(ascending=True)
print(sorted_series)
```

Here, the data is sorted from fewest reports to most reports. However, we can only see the first five of the neighborhoods with fewest reports. 

```python
print(sorted_series.head(20))
```

Now we have the neighborhoods with the fewest reports. Let's also do the neighborhoods with the most reports, just incase it is needed to be cross referenced later.

```python
print(sorted_series.tail(20))
```

===============================================================================================================================

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(18, 4))
sorted_series.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title("Criminal Reports Across Neighborhoods of Pittsburgh")
plt.xlabel('Neighborhood')
plt.ylabel("# of Reports")
plt.show()
```

===============================================================================================================================

Now let's assign each neighborhood a "goodness value." First, we will find the average criminal reports. Then, by comparing each neighborhood, we can determine where each stands from this metric.

```python
total = 0
for index in dictionary:
    total = total + dictionary[index]
avgCrime = total / len(dictionary)
print (avgCrime, "reports per neighborhood")
```

Now, we find the standard deviation.

```python
import statistics
justNums = []
for index in dictionary:
    justNums.append(dictionary[index])
stDev = statistics.stdev(justNums)
rankedDictionary = dictionary
for test in rankedDictionary:
    rankedDictionary[test] = (-(rankedDictionary[test] - avgCrime) / stDev)
sortedRanked = pd.Series(rankedDictionary)
sortedRanked = sortedRanked.sort_values(ascending=True)
print(sortedRanked)
```
